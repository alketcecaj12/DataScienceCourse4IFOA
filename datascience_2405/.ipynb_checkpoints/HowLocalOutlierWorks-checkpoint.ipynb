```python
import numpy as np
from sklearn.neighbors import LocalOutlierFactor
```

```python
X = [[-1.1], [0.2], [101.1], [0.3], [1.9], [0.8], [125.1]]
X
```

```python
clf = LocalOutlierFactor(n_neighbors=3)
clf.fit_predict(X)
```

```python
outliers = clf.negative_outlier_factor_
```
