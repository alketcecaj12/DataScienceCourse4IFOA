### Example of how to use Cross validation on a classification problem

##### The points that this notebook covers are :

- 5-Fold validation 


- 10-Fold validation


-  Bias and variance


- Grid searching the best parameters of the model 

 

#### Load the dataset

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
X = iris.data
y = iris.target
```

```python
y
```

```python
type(y)
```

#### Split the dataset into training set and testing set

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, random_state=1)
```

```python
X_test.shape
```

### Classification algorithm KNN 

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors=35)
```

```python
knn.fit(X_train, y_train)
```

```python
previsione = knn.predict(X_test)
```

```python
previsione
```

```python
from sklearn import metrics
```

```python
metrics.accuracy_score(previsione, y_test)
```

1- is this result too good to be true ? 

2- was I simply lucky ? 

3- is the data somehow biased ? 

### Cross validation 5-Fold version

K = 5

```python
from sklearn.model_selection import cross_val_score
```

```python
k_fold = 5
score = cross_val_score(knn, X, y, cv = k_fold, scoring="accuracy")
```

```python
score
```

```python
bias = score.mean()
print(bias)
```

```python
variance = score.std()
print(variance)
```

### Cross validation 10-Fold version

- K = 10

```python
k_fold = 10
score = cross_val_score(knn, X, y, cv = k_fold, scoring="accuracy")
```

```python
score
```

#### Compute and display bias and variance

```python
bias = score.mean()
print(bias)
```

```python
variance = score.std()
print(variance)
```

#### Use of GridSearchCV class for fine tuning the model 

```python
from sklearn.model_selection import GridSearchCV
```

##### Define a range where to vary the parameters of the model

```python
k_range = list(range(1, 31))
```

```python
print(k_range)
```

```python
param_grid = dict(n_neighbors = k_range)
```

#### Declare the GridSearchCV instance and start using it with a given cross-validation parameter 

```python
k_fold = 5
grid = GridSearchCV(knn, param_grid, cv = k_fold, scoring="accuracy")
```

```python
grid
```

### Fit the GridSearchCV in order to find the best parameter/parameters 

```python
grid.fit(X, y)
```

### Transform the results of GridSearchCV into a dataframe and sort by bias

```python
import pandas as pd
data_grid = pd.DataFrame(grid.cv_results_)
```

```python
data_grid
```

### All what we need from the GridSearchCV results are these four fields i.e. 

- best parameter for the classification model


- the bias 


- the variance 


- the rank of the test score

```python
dati_scoring = data_grid[["params", "mean_test_score", "std_test_score", "rank_test_score"]]
```

```python
dati_scoring.sort_values("rank_test_score")
```
