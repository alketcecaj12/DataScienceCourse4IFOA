```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.DataFrame({"customer1":[1, 1, 1, 0, 0, 0, 1, 1, 1],
                     "customer2":[0, 0, 1, 1, 0, 0, 1, 1, 1], 
                     "customer3":[1, 1, 0, 1, 1, 0, 1, 1, 1],
                     "customer4":[0, 2, 4, 8, 10, 11, 12, 43, 32], 
                     "customer5":[1, 2, 3, 7, 8, 9, 10, 33, 22],
                     "customer6":[0, 2, 4, 8, 10, 11, 12, 43, 32], 
                     "customer7":[1, 2, 3, 7, 8, 9, 10, 33, 22],
                     "customer8":[0, 2, 4, 8, 10, 11, 12, 43, 32], 
                     "customer9":[20,22, 3, 7, 87, 9, 10, 33, 222],
                     "customer10":[19, 22, 4, 8, 100, 11, 12, 43, 322]})
```

```python
data2 = data.T
```

```python
data2.columns = ["f1", "f2", "f3", "f4", "f5", "f6", "f7", "f8", "f9"]
```

```python
data2.index
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
samples = data2.iloc[:, 1:].values
customers = data2.index.tolist()
```

```python
customers
```

```python
from sklearn.preprocessing import normalize 
```

```python
normalized_samples = normalize(samples)
```

```python
plt.figure(figsize = (20, 10))
merge = linkage(normalized_samples, method = "complete")
dendrogram(merge, labels = customers, leaf_rotation=90, leaf_font_size= 6)
plt.xticks(fontsize= 14)
plt.show()
```
