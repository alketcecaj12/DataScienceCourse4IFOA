```python
from sklearn_extra.cluster import KMedoids
import pandas as pd
```

```python
data = pd.DataFrame({"x": [2, 3, 4, 10, 11, 20, 222], 
                    "y":  [1, 2, 1, 17, 18, 29, 282]})
```

```python
data
```

```python
kmedoids = KMedoids(n_clusters=4)
kmedoids.fit(data)
```

```python
kmedoids.labels_
```

```python
kmedoids.cluster_centers_
```

```python
kmedoids.predict([[0,0], [4,4]])
```

```python
kmedoids.inertia_
```

```python
labels = list(kmedoids.labels_)
```

```python
data["lables"] = labels
```
