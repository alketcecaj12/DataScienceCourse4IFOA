```python
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.cluster import KMeans
import numpy as np
```

```python
data = pd.DataFrame({"X": [1, 3, 1, 15, 16, 14, 30, 31, 32], 
                     "Y": [0, 2, 3, 14, 17, 19,  32, 33, 36]})
```

```python
data
```

```python
plt.scatter(data["X"], data["Y"])
plt.show()
```

```python
points = data.values
```

```python
type(points)
```

```python
model = KMeans(n_clusters = 3)
```

```python
model.fit(points)
```

```python
model.inertia_
```

```python
ks = range(1, 8)
in_s = []
for i in ks: 
    model = KMeans(n_clusters=i)
    model.fit(points)
    
    in_i = model.inertia_
    in_s.append(in_i)
    
```

```python
in_s
```

```python
plt.plot(ks, in_s, '-o')
plt.xlabel("numero dei cluster, k")
plt.ylabel("Errore inertia_")
plt.show()
```
