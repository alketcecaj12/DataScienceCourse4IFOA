```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.DataFrame({"x": [1, 1, 1, 2, 2, 2, 6, 7, 7, 11, 12, 12], "y":[1, 2, 2,4, 5, 5, 7, 7, 8, 8, 8, 8]})
```

```python
data
```

```python
from sklearn.cluster import KMeans
```

```python
points = data.values
```

```python
model = KMeans(n_clusters = 3)
```

```python
model.fit(points)
```

```python
model.inertia_
```

```python
ks = range(1, 8)
in_s = []
for i in ks: 
    model = KMeans(n_clusters=i)
    model.fit(points)
    
    in_i = model.inertia_
    in_s.append(in_i)
```

```python
plt.plot(ks, in_s, '-o')
plt.xlabel("numero dei cluster, k")
plt.ylabel("Errore inertia_")
plt.show()
```
