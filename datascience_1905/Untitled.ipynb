```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np
```

```python
exp = np.array([1,     1,   5,   12,   15,   2,   3,   4,   8,   11,     25])
sal = np.array([130, 120, 550, 1400, 1800, 280, 400, 500, 800, 1400,   2955])
```

```python
plt.scatter(exp, sal)
plt.show()
```

```python
sal = sal.reshape(-1, 1)
exp = exp.reshape(-1, 1)
```

```python
np.corrcoef(sal, exp)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(exp, sal)
```

```python
model.predict([[25]])
```
