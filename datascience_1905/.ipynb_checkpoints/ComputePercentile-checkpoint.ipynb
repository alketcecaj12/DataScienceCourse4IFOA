```python
import pandas as pd
```

```python
data = pd.DataFrame({'id':[1000, 1001, 1234, 1254, 1237, 1212, 1211, 1111, 1892, 8901, 8902], 
                     'recent':[2, 3,    4,   28,   29,  201,  205,  300,  301, 600, 601]})
```

```python
data
```

```python
r = list(range(5, 0, -1))
recency_quartiles = pd.qcut(data['recent'], q = 5, labels=r )
```

```python
recency_quartiles
```

```python
data['recency_quartile'] = recency_quartiles
```

```python

data
```

```python
# building real lables
labels = ['active', 'lapsed', 'inactive', 'chuned', 'dead']
recency_quartiles = pd.qcut(data['recent'], q = 5, labels=labels)
data['recency_quartile_labels'] = recency_quartiles
```

```python
data.sort_values('recency_quartile')
```
