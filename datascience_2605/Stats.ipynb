```python
from empiricaldist import Pmf
from empiricaldist import Cdf
import numpy as np
import matplotlib.pyplot as plt
```

### An interval of numbers 

```python
a = np.array([1, 2, 3, 4, 4, 4, 4, 5, 6, 7])
```

#### Compute the probability mass function

```python
pmf = Pmf.from_seq(a)
```

```python
pmf
```

```python
plt.plot(pmf)
plt.show()
```

```python
cdf_a = Cdf.from_seq(a)
```

```python
plt.plot(cdf_a)
```
